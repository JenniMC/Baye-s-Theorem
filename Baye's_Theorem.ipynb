# Probability

## 1.1 Sample Space
The set of all possible outcomes of a statistical experiment is called the sample space and is represented by the symbol $S$.

Each outcome is called an element of the sample space.

## 1.2 Events
1. Event: An event is a subset of a sample space and is denoted by a capital letter.

2. Complement of an element: The complemet of an event $A$ with respect to $S$ is the set of all elements of $S$ that are not in $A$. We denote the complement of $A$ by the symbol $A'$ or $A^c$.

3. Intersection: The intersection of two events $A$ and $B$, represented by the symbol $A\cap B$ is the event that contains all the elements common to $A$ and $B$.

4. Mutually exclusive: Two events $A$ and $B$ are mutually exclusive or disjoint if $A\cap B=\varnothing$, that is, if $A$ and $B$ have not elements in common.

5. Union: The union of two events $A$ and $B$, which is represented by the symbol $A\cup B$, is the event that contains all the elements that belong to $A$ or to $B$, or to both.

## 1.3 Probability of an Event
The probability of an event $A$ is the sum of the weights of all the sample points of $A$. Therefore,
$$0 ≤ P(A) ≤ 1, P(\varnothing) = 0, P(S) = 1$$

Theorem 1. If an experiment can have any $N$ different equally likely outcomes, and if exactly $n$ of these outcomes correspond to event $A$, then the probability of the latter is:
$$P(A) = n/N$$

## 1.4 Additive Rules
Theorem 2. If $A$ and $B$ are any two events, then
$$P(A ∩ B) = P(A) + P(B) - P(A ∩ B)$$

### Corollary
1. If $A$ and $B$ are mutually exclusive, then
$$P(A ∩ B) = P(A) + P(B)$$
2. If $A_1, A_2, A_3, ..., A_n$ are mutually exclusive, then
$$P(A_1 ∩ A_2 ∩ ... ∩ A_n) = P(A_1) + P(A_2) + ... + P(A_n)$$
3. If $A_1, A_2, A_3, ..., A_n$ is a partition of a sample space $S$, then
$$P(A_1 ∩ A_2 ∩ ... ∩ A_n)=P(A_1)+P(A_2)+...+P(A_n)=P(S)=1$$

Theorem 3. For three events $A$, $B$ and $C$
$$P(A\cup B\cup C)=P(A)+P(B)+P(C)-P(A\cap B)-P(A\cap C)-P(B\cap C)+P(A\cap B\cap C)$$

Theorem 4. If $A$ and $A^c$ are complementary events, then
$$P(A)+P(A^c)=1$$

## 1.5 Conditional Probability
The conditional probability of $B$, given $A$, denoted by $P(B|A)$, is defined as follows:
$$P(B|A)=\frac{P(A\cap B)}{P(A)}$$
if $P(A)>0$

Two events $A$ and $B$ are independent if and only if:
$$P(B|A)=P(B)$$
and
$$P(A|B)=P(A)$$
Otherwise, $A$ and $B$ are dependent.

## 1.6 Multiplicative Rules
Theorem 5. If events $A$ and $B$ can occur in an experiment, then
$$P(A ∩ B) = P(A)P(B ∩ A)$$

Theorem 6. Two events $A$ and $B$ are independent if and only if
$$P(A ∩ B) = P(A)P(B)$$

# Baye's Theorem
Theorem 7. If the events $B_1, B_2, ..., B_k$ constitute a subdivision of the sample space $S$, such that $P(B_i) ≥ 0$ for $i = 1, 2, ..., k$, then for any event $A$ in $S$:
$$P(A) = \sum_{i=1}^k P(B_i ∩ A) = \sum_{i=1}^k P(B_i) P(A | B_i)$$

Theorem 8. **Bayes' Theorem**

If the events $B_1, B_2, ..., B_k$ constitute a subdivision of the sample space $S$, where $P(B_i) ≥ 0$ for $i = 1, 2, ..., k$, then for any event $A$ in $S$, $P(A) ≥ 0$

$$P(B_r|A)=\frac{P(B_r\cap A)}{\sum^k_{i=1}P(B_i\cap A)}=\frac{P(B_r)P(A|B_r)}{\sum^k_{i=1}P(B_i)P(A|B_i)}$$
for $r=1,2,...,k$

## Problem
It is estimated that 0.0001% of the population has a certain disease. A test to detect this disease has a 99% accuracy rate for positive cases and a 95% accuracy rate for negative cases. If a randomly selected person tests positive, what is the probability that they actually have the disease?

The percentage of the population with the disease is 0.0001, therefore the percentage of the population without the disease is 0.9999.

A1 has the disease (0.0001).

A2 does not have the disease (0.9999).

```python
A1 = 0.0001
A2 = 0.9999
```

The test is 99% effective at detecting the disease.

BA1 The test is positive when the person has the disease .99

```python
BA1= 0.99
```

The test is 95% effective in confirming a negative result when the person does not have the disease.

CA2: The test is negative when the person does not have the disease .95

```python
CA2 = 0.95
```

So, the probability of testing positive and not having the disease, BA2

```python
BA2=1-CA2
BA2
```

So, the probability of testing negative when you have the disease, CA1

```python
CA1=1-BA1
CA1
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
previous_prob = [A1, A2]
previous_labels = ['P(Disease) = A1', 'P(Not Disease) = A2']
plt.bar(previous_labels, previous_prob, color=['skyblue', 'lightcoral'])
plt.title('Previous Probabilities of the Disease')
plt.ylabel('Probability')
plt.ylim(0, 1)
plt.text(0, A1 + 0.00005, f'{A1:.4f}', ha='center', va='bottom', fontsize=10)
plt.text(1, A2 - 0.05, f'{A2:.4f}', ha='center', va='top', fontsize=10)

plt.subplot(1, 2, 2)

prob_given_disease = [BA1, CA1]
labels_given_disease = ['P(Positive|A1)', 'P(Negative|A1)']

prob_given_not_disease = [BA2, CA2]
labels_given_not_disease = ['P(Positive|A2)', 'P(Negative|A2)']

x = np.arange(len(labels_given_disease))
width = 0.35

rects1 = plt.bar(x - width/2, prob_given_disease, width, label='If you have the disease', color='skyblue')
rects2 = plt.bar(x + width/2, prob_given_not_disease, width, label='If you DO NOT have the disease', color='lightcoral')

plt.ylabel('Probability')
plt.title('Conditional Probabilities of the Test')
plt.xticks(x, ['Positive Result', 'Negative Result'])
plt.legend()
plt.ylim(0, 1.1)

def autolabel(rects, values):
    for rect, value in zip(rects, values):
        height = rect.get_height()
        plt.annotate(f'{value:.2f}',
                     xy=(rect.get_x() + rect.get_width() / 2, height),
                     xytext=(0, 3),
                     textcoords="offset points",
                     ha='center', va='bottom', fontsize=10)

autolabel(rects1, prob_given_disease)
autolabel(rects2, prob_given_not_disease)

plt.tight_layout()
plt.show()
```

The full probability when the test is positive PB

```python
PB = (BA1*A1) + (BA2*A2)
PB
```

The complete probability when the test is negative PN

```python
PN=(CA1*A1)+(CA2*A2)
PN
```

Therefore,
1. PA1B: The probability of having the disease when testing positive
2. PA2B: The probability of NOT having the disease when testing positive
3. PA1C: The probability of having the disease when testing negative
4. PA2C: The probability of NOT having the disease when testing negative

```python

PA1B = (A1*BA1) / PB
print("The probability of having the disease when testing positive is:",PA1B)
```

```python
PA2B = (A2*BA2) / PB
print("The probability of NOT having the disease when testing positive is:",PA2B)
```

The sum of the probabilities when it is positive PBT

```python
PBT=PA1B+PA2B
PBT
```

```python
PA1C = (A1*CA1) / PN
print("The probability of having the disease when testing negative is:",PA1C)
```

```python
PA2C=(A2*CA2)/PN
print("The probability of NOT having the disease when testing negative is:",PA2C)
```

The sum of the probabilities when it is negative PNT

```python
PNT=PA1C+PA2C
PNT
```
